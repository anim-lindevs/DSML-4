```python
import matplotlib.pyplot as plt
import numpy as np

# ১. 1D অ্যারে তৈরি
x = np.linspace(0, 10, 100)
y = np.linspace(0, 10, 100)

# ২. 2D মেসগ্রিড তৈরি
X, Y = np.meshgrid(x, y)  # meshgrid এর মাধ্যমে দুই মাত্রা তৈরি সম্ভব :contentReference[oaicite:3]{index=3}

# ৩. Z নির্ধারণ
Z = np.sin(X) * np.cos(Y)

# ৪. গ্রাফ প্লট করা
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='plasma', edgecolor='k', linewidth=0.3)
ax.set_title("3D folded origami pattern")
plt.show()
```

```python
import plotly.express as px

country = input("Enter the country name: ")
data = {
    'Country': [country],
    'Values':  [100]
}

fig = px.choropleth(
    data_frame=data,
    locations='Country',
    locationmode='country names',
    color='Values',
    color_continuous_scale='Inferno',
    scope='world',  # নিশ্চিতভাবে বিশ্বজুড়ে দেখাবে
    title=f'Country Map Highlighting {country}'
)

fig.update_layout(
    coloraxis_colorbar=dict(title='Value', thickness=10, len=0.6)
)
fig.update_geos(projection_type="natural earth")

fig.show()
```
