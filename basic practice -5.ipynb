```python
# write a function that checks if a prime number 
# right function 
import math

def is_prime(num):
    if num <= 1:
        return False  # 1 বা তার কম সংখ্যা মৌলিক নয়
    if num == 2:
        return True   # 2 একটি মৌলিক সংখ্যা
    if num % 2 == 0:
        return False  # জোড় সংখ্যা মৌলিক নয়

    for i in range(3, int(math.sqrt(num)) + 1, 2):
        if num % i == 0:
            return "not prime number"
    return "prime number"


is_prime(19)
is_prime(21)
```
