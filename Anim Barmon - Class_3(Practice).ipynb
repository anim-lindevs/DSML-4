###`Problem 1:` Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

````python
# Write code here

list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

combined_list = [list(pair) for pair in zip(list1, list2)]

longer_list = list1 if len(list1) > len(list2) else list2
combined_list.extend([[item] for item in longer_list[len(combined_list):]])

print(combined_list)
````

###`Problem 3:` Update no of items available

Suppose you are given a list of candy and another list of same size representing no of items of respective candy.

i.e -  
```
candy_list = ['Jelly Belly','Kit Kat','Double Bubble','Milky Way','Three Musketeers']
no_of_items = [10,20,34,74,32]
```

Write a program to show no. of items of each candy type.

`Output:`

```
Jelly Belly-10
Kit Kat-20
Double Bubble-34
Milky Way-74
Three Musketeers-32

```

````python
# Write code here

candy_list = ['Jelly Belly','Kit Kat','Double Bubble','Milky Way','Three Musketeers']
no_of_items = [10, 20, 34, 74, 32]

for candy, count in zip(candy_list, no_of_items):
    print(f"{candy}-{count}")
````

###`Problem 4:` Running Sum on list
Write a program to print a list after performing running sum on it.

i.e:

`Input:`
```
list1 = [1,2,3,4,5,6]
```
`Output:`
```
[1,3,6,10,15,21]
```

````python
# Write code here

list1 = [1, 2, 3, 4, 5, 6]

running_sum = [sum(list1[:i+1]) for i in range(len(list1))]

print(running_sum)
````

###`Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
# Write code here

list1 = [2, 4, 6, 10, 1]

result = []

for i in list1:
    total = i + sum(x for x in list1 if x > i)
    result.append(total)

print(result)
````

###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
# Write code here

num1 = [23, 45, 67, 78, 89, 34]
num2 = [34, 89, 55, 56, 39, 67]

common = list(set(num1) & set(num2))
common.sort()

print(common)
````


### `Problem 9:` Convert a character matrix to a single string using string comprehension.

**Example 1:**

Input:
```bash
[['t', 'h', 'i', 's'], ['i', 's'], ['a'], ['t', 'e', 's', 't']]
```

Output:
```bash

```

````python
# Write code here

matrix = [['t', 'h', 'i', 's'], ['i', 's'], ['a'], ['t', 'e', 's', 't']]

result = ' '.join(''.join(chars) for chars in matrix)

print(result)
````

### `Problem 11:` Write a program that can find the max number of each row of a matrix

**Example:**

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

````python
# Write code here
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

max_in_rows = [max(row) for row in matrix]

print(max_in_rows)
````

### `Problem 13:` Write a list comprehension to print the following matrix

[[0, 1, 2], [3, 4, 5], [6, 7, 8]]

````python
# Write code here

matrix = [[3 * i + j for j in range(3)] for i in range(3)]

print(matrix)
````

### Problem 14: Extract Even Numbers from a List

**Task:** Given a list of integers, use list comprehension to create a new list containing only the even numbers.

**Example:**

Input:
```bash
[1, 2, 3, 4, 5, 6, 7, 8, 9]
```

Output:
```bash
[2, 4, 6, 8]
```


````python
nums = [1, 2, 3, 4, 5, 6, 7, 8, 9]

even_nums = [num for num in nums if num % 2 == 0]

print(even_nums)
````

### Problem 15: Convert Strings to Uppercase

**Task:** Given a list of strings, use list comprehension to convert all strings to uppercase.

**Example:**

Input:
```bash
['apple', 'banana', 'cherry']
```

Output:
```bash
['APPLE', 'BANANA', 'CHERRY']
```


````python
fruits = ['apple', 'banana', 'cherry']

uppercase_fruits = [fruit.upper() for fruit in fruits]

print(uppercase_fruits)
````
