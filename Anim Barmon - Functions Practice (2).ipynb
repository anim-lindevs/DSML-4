### **`Problem-1:`** Write a Python function that takes a list and returns a new list with unique elements of the first list.

**Exercise 1:**

Input:

```bash
[1,2,3,3,3,3,4,5]
```

Output:

```bash
[1, 2, 3, 4, 5]
```

````python
def unique_list(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result
print(unique_list([1, 2, 3, 3, 3, 3, 4, 5]))

````

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
# Write code here

def sort_hyphenated(sequence: str) -> str:
    parts = sequence.split('-')
    parts.sort()
    return '-'.join(parts)

s = "green-red-yellow-black-white"
print(sort_hyphenated(s))
````

### **`Problem 3:`** Write a Python function that counts uppercase and lowercase letters in a string.

**Sample Input:**  
```python
'The Quick Brown Fox Jumps Over The Lazy Dog.'
```

**Expected Output:**  
```
No. of Upper case characters: 8  
No. of Lower case characters: 26
```

````python
# Write code here

def count_case_chars(s: str) -> tuple[int, int]:
    upper = sum(1 for c in s if c.isupper())
    lower = sum(1 for c in s if c.islower())
    return upper, lower

text = 'The Quick Brown Fox Jumps Over The Lazy Dog.'
u, l = count_case_chars(text)
print(f"No. of Upper case characters: {u}")
print(f"No. of Lower case characters: {l}")
````

### **`Problem 4:`** Write a Python program to print the even numbers from a given list.
```
Sample List : [1, 2, 3, 4, 5, 6, 7, 8, 9]
Expected Result : [2, 4, 6, 8]
```

````python
# Write code here
def get_evens(lst):
    return [x for x in lst if x % 2 == 0]

print(get_evens([1, 2, 3, 4, 5, 6, 7, 8, 9]))

````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here

import math

def is_perfect(n: int) -> bool:
    if n < 2:
        return False
    total = 1 

    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            total += i
            j = n // i
            if j != i:
                total += j
    return total == n

for num in [6, 28, 496, 8128, 12]:
    print(f"{num} is perfect? {is_perfect(num)}")
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
# Write code here
def concat_dicts(*dicts):
    result = {}
    for d in dicts:
        result.update(d)
    return result
dic1 = {1:10, 2:20}
dic2 = {3:30, 4:40}
dic3 = {5:50, 6:60}
print(concat_dicts(dic1, dic2, dic3))

````

### `Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
# Write code here

from collections import Counter

def most_frequent_word(text: str) -> tuple[str, int]:
    words = text.split()
    counts = Counter(words)
    word, freq = counts.most_common(1)[0]
    return word, freq

input_text = 'hello how are you i am fine thank you'
word, freq = most_frequent_word(input_text)
print(f"{word} -> {freq}")
````

###`Problem-8` Using filter() and list() functions and .lower() method filter all the vowels in a given string.



````python
# Write code here

def filter_vowels(s: str) -> list[str]:
    return list(filter(lambda c: c.lower() in 'aeiou', s))

text = "Hello World!"
print(filter_vowels(text))
````

`Problem-9`: Use reduce to convert a 2D list to 1D

````python
# Write code here
from functools import reduce

def flatten_2d_to_1d(lst):
    return reduce(lambda x, y: x + y, lst)

input_list = [[1, 2, 3], [4, 5], [6, 7]]
flattened_list = flatten_2d_to_1d(input_list)
print(flattened_list)
````

`Problem 10`- A dictionary contains following information about 5 employees:
- First name
- Last name
- Age
- Grade(Skilled,Semi-skilled,Highly skilled)<br>
Write a program using map/filter/reduce to a list of employees(first name + last name) who are highly skilled

````python
# Write code here

from functools import reduce
employees = [
    {'first_name': 'Alice', 'last_name': 'Smith', 'age': 30, 'grade': 'Skilled'},
    {'first_name': 'Bob', 'last_name': 'Brown', 'age': 40, 'grade': 'Highly skilled'},
    {'first_name': 'Charlie', 'last_name': 'Davis', 'age': 35, 'grade': 'Semi-skilled'},
    {'first_name': 'David', 'last_name': 'Evans', 'age': 50, 'grade': 'Highly skilled'},
    {'first_name': 'Eve', 'last_name': 'Garcia', 'age': 28, 'grade': 'Skilled'}
]

highly_skilled = filter(lambda emp: emp['grade'] == 'Highly skilled', employees)
full_names = map(lambda emp: f"{emp['first_name']} {emp['last_name']}", highly_skilled)
result = list(full_names)
print(result)
````
