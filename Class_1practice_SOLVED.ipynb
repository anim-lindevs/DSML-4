### Q2:- Write a program that will convert celsius value to fahrenheit.

````python
# Take Celsius input from the user
n = input("Enter the Celsius value: ")

# Check if the input is a valid number (integer or float)
try:
    m = float(n)  # Convert input to float to handle decimal values
    f = (m * 9/5) + 32
    print(f"{m}°C is equal to {f}°F.")  # Clear and informative output
except ValueError:
    print("Invalid input! Please enter a valid number.")
````

For example:
Input:
heads -> 4

<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
# Write your code here
heads = int(input("Enter the number of heads: "))
legs = int(input("Enter the number of legs: "))

if heads > 0 and legs > 0:
    if legs % 2 == 0 and 2 * heads <= legs <= 4 * heads:
        chickens = (4 * heads - legs) // 2
        dogs = heads - chickens
        print("Number of chickens:", chickens)
        print("Number of dogs:", dogs)
    else:
        print("Invalid number of legs for the given number of heads.")
else:
    print("Heads and legs must be positive numbers.")
````

### Q7:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Taking user input
a = int(input("Enter the first term: "))
s_t = int(input("Enter the second term: "))
n = int(input("Enter the position of the term you want to find (n): "))

# Ensure n is a positive number
if n <= 0:
    print("Please enter a positive number for the term position.")
else:
    # Calculate the common difference
    d = s_t - a

    # Calculate the nth term
    nth_term = a + (n - 1) * d

    # Display the result
    print(f"The {n}th term of the arithmetic series is: {nth_term}")
````

### Q8:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# Taking input for the two fractions
try:
    num1 = int(input("Enter numerator of the first fraction: "))
    den1 = int(input("Enter denominator of the first fraction: "))
    num2 = int(input("Enter numerator of the second fraction: "))
    den2 = int(input("Enter denominator of the second fraction: "))

    # Check if the denominators are zero
    if den1 == 0 or den2 == 0:
        print("Denominator cannot be zero. Please enter valid fractions.")
    else:
        # Calculate the sum of the two fractions
        common_den = den1 * den2
        numerator_sum = (num1 * den2) + (num2 * den1)

        # Print the result as a fraction
        print(f"The sum of the fractions is: {numerator_sum}/{common_den}")
except ValueError:
    print("Invalid input! Please enter valid integers for numerators and denominators.")
````

### Q9:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
import math

# Dimensions of the milk tank (in cm)
H = float(input("Enter the height of the tank (cm): "))
L = float(input("Enter the length of the tank (cm): "))
B = float(input("Enter the breadth of the tank (cm): "))

# Dimensions of the glass
h = float(input("Enter the height of the glass (cm): "))
r = float(input("Enter the radius of the glass (cm): "))

# Calculate the volume of the milk tank (in cubic cm)
volume_of_tank = H * L * B

# Calculate the volume of one glass (in cubic cm)
volume_of_glass = math.pi * (r ** 2) * h

# Calculate the number of glasses
number_of_glasses = volume_of_tank / volume_of_glass

# Output the result
print(f"The number of glasses of milk that can be obtained: {int(number_of_glasses)}")
````
